# Polarização e Variância

Seja o modelo ARX $y[k] = a_1y[k-1] + a_2y[k-2] + b_1u[k-1] + b_2u[k-2] + e[k]$ e o ruído branco gaussiano $\nu[k] \sim \mathcal{N}(0, \sigma^2)$.

## Regressores gerados por ARX

Para os regressores $u, y$ obtidos por $y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + \nu[k]$, são esperados:

### Polarização

Vale

\begin{equation*}
Y = \Psi\Theta + \nu
\end{equation*}

que, estendida, é

\begin{equation*}
\begin{bmatrix}
y[2] \\
y[3] \\
y[4] \\
\vdots \\
y[N]
\end{bmatrix} =
\begin{bmatrix}
y[1] & y[0] & u[1] & u[0] \\
y[2] & y[1] & u[2] & u[1] \\
y[3] & y[2] & u[3] & u[2] \\
\vdots & \vdots & \vdots & \vdots \\
y[N-1] & y[N-2] & u[N-1] & u[N-2]
\end{bmatrix}
\begin{bmatrix}
a_1 \\
a_2 \\
b_1 \\
b_2
\end{bmatrix} +
\begin{bmatrix}
\nu[2] \\
\nu[3] \\
\nu[4] \\
\vdots \\
\nu[N]
\end{bmatrix} \\
\end{equation*}

Assumimos $E[u\nu] = 0$. Para o passo $k$, os regressores $y[k - 1]$ e $y[k - 2]$ são correlacionados com os termos de ruído passado $\nu[k - 1]$ e $\nu[k - 2]$, mas não com o respectivo termo de ruído atual $\nu[k]$. Assim, $E[y[k - l]\nu[k]] = 0$ e, portanto, $E[\Psi^T\nu] = 0$, que atende ao critério para viés nulo $E[Ae] = 0$.

### Variância

Dado que o estimador é não polarizado para a estrutura ARX, a variância do estimador é dada por

\begin{equation*}
\text{cov}[\hat{\Theta}] = E[(\Psi^{T}\Psi)^{-1}]\sigma_e^2
\end{equation*}

Considerando as condições, mais comuns, nas quais não $\sigma_e^2$ não é conhecido, a variância do erro de medição é estimada pela variância dos resíduos, $\sigma_\xi^2$.

Assim, a variância estimada é:

\begin{equation*}
\text{cov}[\hat{\Theta}] = E[(\Psi^{T}\Psi)^{-1}]\sigma_\xi^2
\end{equation*}

## Regressores gerados por ARMAX

Para os regressores $u, y$ obtidos por $y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + \nu[k] - 0.8\nu[k-1]$, são esperados:

### Polarização

Vale

\begin{equation*}
Y = \Psi\Theta + e
\end{equation*}

que, estendida, é

\begin{equation*}
\begin{bmatrix}
y[2] \\
y[3] \\
y[4] \\
\vdots \\
y[N]
\end{bmatrix} =
\begin{bmatrix}
y[1] & y[0] & u[1] & u[0] \\
y[2] & y[1] & u[2] & u[1] \\
y[3] & y[2] & u[3] & u[2] \\
\vdots & \vdots & \vdots & \vdots \\
y[N-1] & y[N-2] & u[N-1] & u[N-2]
\end{bmatrix}
\begin{bmatrix}
a_1 \\
a_2 \\
b_1 \\
b_2
\end{bmatrix} +
\begin{bmatrix}
\nu[2] - 0.8\nu[1] \\
\nu[3] - 0.8\nu[2] \\
\nu[4] - 0.8\nu[3] \\
\vdots \\
\nu[N] - 0.8\nu[N-1]
\end{bmatrix} \\
\end{equation*}

Assumimos $E[u\nu] = 0$. Para esta geradora, no entanto,

\begin{equation*}
y[k - 1] = 1.5y[k - 2] - 0.7y[k - 3] + 1.0u[k - 2] + 0.5u[k - 3] + \nu[k - 1] - 0.8\nu[k - 2]
\end{equation*}

e o respectivo erro,

\begin{equation*}
e[k] = \nu[k] - 0.8\nu[k - 1]
\end{equation*}

Portanto, $E[y[k - 1]e[k]] \neq 0$ e, consequentemente, $E[\Psi^Te] \neq 0$, o que não atende ao critério para viés nulo $E[Ae] = 0$. Assim, o estimador é polarizado.

### Variância

Dado que o estimador é polarizado para a estrutura ARMAX, a variância do estimador é dada por

\begin{equation*}
\text{cov}[\hat{\Theta}] = E[\bm{\hat{\Theta}}\bm{\hat{\Theta}}^T] - E[\bm{\hat{\Theta}}]E[\bm{\hat{\Theta}}]^T
\end{equation*}

## Regressores gerados por OE

Para os regressores $u, y$ obtidos por $y[k] = y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + \nu[k] - 1.5\nu[k-1] + 0.7\nu[k-2]$, são esperados:

### Polarização

Vale

\begin{equation*}
Y = \Psi\Theta + e
\end{equation*}

que, estendida, é
\begin{equation*}
\begin{bmatrix}
y[2] \\
y[3] \\
y[4] \\
\vdots \\
y[N]
\end{bmatrix} =
\begin{bmatrix}
y[1] & y[0] & u[1] & u[0] \\
y[2] & y[1] & u[2] & u[1] \\
y[3] & y[2] & u[3] & u[2] \\
\vdots & \vdots & \vdots & \vdots \\
y[N-1] & y[N-2] & u[N-1] & u[N-2]
\end{bmatrix}
\begin{bmatrix}
a_1 \\
a_2 \\
b_1 \\
b_2
\end{bmatrix} +
\begin{bmatrix}
\nu[2] - 1.5\nu[1] + 0.7\nu[0] \\
\nu[3] - 1.5\nu[2] + 0.7\nu[1] \\
\nu[4] - 1.5\nu[3] + 0.7\nu[2] \\
\vdots \\
\nu[N] - 1.5\nu[N-1] + 0.7\nu[N-2]
\end{bmatrix} \\
\end{equation*}

Assumimos $E[u\nu] = 0$. Para esta geradora, no entanto,

\begin{equation*}
y[k - 1] = 1.5y[k - 2] - 0.7y[k - 3] + 1.0u[k - 2] + 0.5u[k - 3] + \nu[k - 1] - 1.5\nu[k - 2] + 0.7\nu[k - 3]
\end{equation*}

e o respectivo erro,

\begin{equation*}
e[k] = \nu[k] - 1.5\nu[k - 1] + 0.7\nu[k - 2]
\end{equation*}

Portanto, $E[y[k - 1]e[k]] \neq 0$ e, consequentemente, $E[\Psi^Te] \neq 0$, o que não atende ao critério para viés nulo $E[Ae] = 0$. Assim, o estimador é polarizado.

### Variância

Dado que o estimador é polarizado para a estrutura OE, a variância do estimador é dada por

\begin{equation*}
\text{cov}[\hat{\Theta}] = E[\bm{\hat{\Theta}}\bm{\hat{\Theta}}^T] - E[\bm{\hat{\Theta}}]E[\bm{\hat{\Theta}}]^T
\end{equation*}

```python
# y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + e[k]
def arx(y, u, e, k):
    if k < 2:
        return e[k]  # Not enough data for y[k-1], y[k-2], u[k-1], u[k-2]
    return 1.5 * y[k - 1] - 0.7 * y[k - 2] + 1.0 * u[k - 1] + 0.5 * u[k - 2] + e[k]


# y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + e[k] - 0.8e[k-1]
def armax(y, u, e, k):
    if k < 2:
        return e[k]  # Not enough data for y[k-1], y[k-2], u[k-1], u[k-2], e[k-1]
    return (
        1.5 * y[k - 1]
        - 0.7 * y[k - 2]
        + 1.0 * u[k - 1]
        + 0.5 * u[k - 2]
        + e[k]
        - 0.8 * e[k - 1]
    )


# y[k] = 1.5y[k-1] - 0.7y[k-2] + 1.0u[k-1] + 0.5u[k-2] + e[k] - 1.5e[k-1] + 0.7e[k-2]
def oe(y, u, e, k):
    if k < 2:
        return e[
            k
        ]  # Not enough data for y[k-1], y[k-2], u[k-1], u[k-2], e[k-1], e[k-2]
    return (
        1.5 * y[k - 1]
        - 0.7 * y[k - 2]
        + 1.0 * u[k - 1]
        + 0.5 * u[k - 2]
        + e[k]
        - 1.5 * e[k - 1]
        + 0.7 * e[k - 2]
    )


import numpy as np

# Set random seed for reproducibility
np.random.seed(0)
N = 1000

# u = white uniform in [-1, 1]
u = np.random.uniform(-1, 1, N)
```

```python
def build_regression_matrix(y, u, na, nb, nk=0):
    """
    Build the regression matrix for ARX model.

    Parameters:
    y : np.ndarray
        Output data.
    u : np.ndarray
        Input data.
    na : int
        Number of poles (order of A).
    nb : int
        Number of zeros (order of B).
    nk : int, optional
        Input delay (default is 0).

    Returns:
    Psi : np.ndarray
        Regression matrix.
    """
    N = len(y)
    assert len(u) == N, "Input and output data must have the same length."

    # Ensure 0 mean for input and output
    u = u - np.mean(u)
    y = y - np.mean(y)

    # Construct the regression matrix
    Psi = np.zeros((N - max(na, nb + nk), na + nb))
    for k in range(max(na, nb + nk), N):
        # Fill in the A part (output past values)
        for i in range(1, na + 1):
            Psi[k - max(na, nb + nk), i - 1] = -y[k - i]
        # Fill in the B part (input past values with delay)
        for j in range(1, nb + 1):
            Psi[k - max(na, nb + nk), na + j - 1] = u[k - j - nk]

    return Psi


def fit_arx(
    y: np.ndarray, u: np.ndarray, na: int, nb: int, nk: int = 0
) -> tuple[np.ndarray, np.ndarray, np.ndarray]:
    """
    Estimate ARX model parameters using least squares.

    Parameters:
    y : np.ndarray
        Output data.
    u : np.ndarray
        Input data.
    na : int
        Number of poles (order of A).
    nb : int
        Number of zeros (order of B).
    nk : int, optional
        Input delay (default is 0).

    Returns:
    A : np.ndarray
        Estimated coefficients of A polynomial.
    B : np.ndarray
        Estimated coefficients of B polynomial.
    Psi : np.ndarray
        Regression matrix used in estimation.
    """
    N = len(y)
    assert len(u) == N, "Input and output data must have the same length."

    # Ensure 0 mean for input and output
    u = u - np.mean(u)
    y = y - np.mean(y)

    # Construct the regression matrix
    Psi = build_regression_matrix(y, u, na, nb, nk)

    # Corresponding output vector
    y_reg = y[max(na, nb + nk) :]

    # Solve the least squares problem
    theta, _, _, _ = np.linalg.lstsq(Psi, y_reg, rcond=None)

    # Separate A and B coefficients
    A = np.concatenate(([1], theta[:na]))  # A[0] = 1
    B = np.concatenate((np.zeros(nk + 1), theta[na:]))  # B[0:nk] = 0

    return A, B, Psi
```

```python
n_tests = 1000


def run_for_system(sys, sys_name):

    As, Bs, Psis, xis = [], [], [], []

    for _ in range(n_tests):
        # nu = white gaussian with zero mean and variance 1
        nu = np.random.normal(0, np.sqrt(2), N)

        # Set initial conditions
        y = np.zeros(N)

        # Simulate the system
        for k in range(N):
            y[k] = sys(y, u, nu, k)

        # Fit ARX model
        A, B, Psi = fit_arx(y, u, na=2, nb=2)

        # Discard the non-estimated leading coefficients
        A, B = A[1:], B[1:]

        # Compute residuals
        y_reg = y[max(2, 2) :]
        xi = y_reg - Psi @ np.concatenate((A, B))

        As.append(A)
        Bs.append(B)
        Psis.append(Psi)
        xis.append(xi)

    As = np.array(As)
    Bs = np.array(Bs)
    Psis = np.array(Psis)
    xis = np.array(xis)

    # Bias
    bias = np.linalg.norm(
        np.mean(np.concatenate((As, Bs), axis=1), axis=0)
        - np.array([-1.5, 0.7, 1.0, 0.5])
    )
    print(f"Bias when the system is {sys_name}: {bias:.3f}")

    # Variance

    if sys_name == "ARX":
        ## \text{cov}[\hat{\Theta}] = E[(\Psi^{T}\Psi)^{-1}]\sigma_\xi^2
        theta_cov_nonbiased = np.mean(
            [np.linalg.inv(Psi.T @ Psi) for Psi in Psis], axis=0
        ) * np.var(xis)
        print(
            f"Variance by E[(Psi^T Psi)^-1] * sigma_xi^2 when the system is {sys_name}:"
        )
        print(np.array2string(theta_cov_nonbiased, precision=1, suppress_small=True))

    ## \text{cov}[\hat{\Theta}] = E[\bm{\hat{\Theta}}\bm{\hat{\Theta}}^T] - E[\bm{\hat{\Theta}}]E[\bm{\hat{\Theta}}^T]
    theta_estimates = np.concatenate((As, Bs), axis=1)
    theta_mean = np.mean(theta_estimates, axis=0)
    theta_cov_biased = np.mean(
        [np.outer(theta, theta.T) for theta in theta_estimates], axis=0
    ) - np.outer(theta_mean, theta_mean.T)
    print(
        f"Variance by E[theta theta^T] - E[theta]E[theta^T] when the system is {sys_name}:"
    )
    print(np.array2string(theta_cov_biased, precision=1, suppress_small=True))

    ## Direct computation
    theta_cov_direct = np.cov(theta_estimates, rowvar=False)
    print(f"Variance by direct computation when the system is {sys_name}:")
    print(np.array2string(theta_cov_direct, precision=3, suppress_small=True))
```

```python
run_for_system(arx, "ARX")
```

```python
run_for_system(armax, "ARMAX")
```

```python
run_for_system(oe, "OE")
```

# Estudo de Estimador para Dados Reais

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data from the file
data = pd.read_csv("bfg33.dat", sep="\\s+", names=["t", "u", "y"])

# Process the data
steps = np.arange(len(data["t"]))
u, y = np.array(data["u"]), np.array(data["y"])

plt.figure(figsize=(12, 6))

plt.subplot(2, 1, 1)
plt.plot(steps, u, color="blue")
plt.title("Entrada BFG33")
plt.xlabel("Amostra")
plt.ylabel("u")
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(steps, y, color="orange")
plt.title("Saída BFG33")
plt.xlabel("Amostra")
plt.ylabel("y")
plt.grid()

plt.tight_layout()
plt.savefig("entrega/imagens/bfg33_entrada_saida.png")
plt.show()
```

```python
# Estimate ARX(na, nb, nk)
na, nb, nk = 2, 2, 20
A, B, Psi = fit_arx(y, u, na=na, nb=nb, nk=nk)


# Compute estimated output
def arx_fitted(y, u, e, k, A, B, na, nb, nk):
    if k < max(na, nb + nk):
        return 0  # Not enough data for past values

    result = 0
    # Add A coefficients (past output values)
    for i in range(1, na + 1):
        result -= A[i] * y[k - i]

    # Add B coefficients (past input values with delay)
    for j in range(1, nb + 1):
        if k - j - nk >= 0:
            result += B[j + nk] * u[k - j - nk]

    return result


# Discard the non-estimated leading coefficients
A_est, B_est = A[1:], B[1:]

# Compute estimated output
y_est = np.zeros_like(y)
for k in range(len(y)):
    y_est[k] = arx_fitted(y, u, np.zeros_like(y), k, A, B, na, nb, nk)

# Plot output comparison
plt.figure(figsize=(15, 3))
plt.plot(steps, y, label="Saída Real", color="orange")
plt.plot(
    steps[max(na, nb + nk) :],
    y_est[max(na, nb + nk) :],
    label="Saída Estimada",
    color="green",
    linestyle="--",
)
plt.text(
    200,
    1.8,
    f"A = [1, {A_est[0]:.2f}, {A_est[1]:.2f}], B = [0, ..., {B_est[0]:.2f}, {B_est[1]:.2f}]",
    fontsize=12,
)

plt.title("Comparação entre Saída Real e Estimada por ARX(2,2,20), BFG33")
plt.xlabel("Amostra")
plt.ylabel("y")
plt.legend()
plt.savefig("entrega/imagens/bfg33_saida_estimada.png")
plt.show()
```

```python
# Estimate covariance assuming non-biased
theta_cov = np.linalg.inv(Psi.T @ Psi) * np.var(
    y[max(na, nb + nk) :] - y_est[max(na, nb + nk) :]
)

# Generate 100 gaussian parameter sets
num_samples = 100
theta_samples = np.random.multivariate_normal(
    np.zeros(len(theta_cov)), theta_cov, num_samples
)
theta_samples += np.concatenate((A_est, B_est[-nb:]))

# Compute estimated outputs for sampled parameters
for i in range(num_samples):
    A_samp = np.concatenate(([1], theta_samples[i, :na]))
    B_samp = np.concatenate((np.zeros(nk + 1), theta_samples[i, na:]))
    y_samp = np.zeros_like(y)
    for k in range(len(y)):
        y_samp[k] = arx_fitted(y, u, np.zeros_like(y), k, A_samp, B_samp, na, nb, nk)
    if i == 0:
        y_samples = y_samp[np.newaxis, :]
    else:
        y_samples = np.vstack((y_samples, y_samp))

# Plot output cloud
plt.figure(figsize=(15, 3))
plt.plot(steps, y, label="Saída Real", color="orange")
for i in range(num_samples):
    plt.plot(
        steps[max(na, nb + nk) :],
        y_samples[i, max(na, nb + nk) :],
        color="gray",
        alpha=0.1,
    )
plt.title("Nuvem de 100 Saídas Estimadas por ARX(2,2,20), BFG33")
plt.xlabel("Amostra")
plt.ylabel("y")
plt.legend()
plt.savefig("entrega/imagens/bfg33_nuvem_saidas.png")
plt.show()
```
